# Step 3 – Process

```r
# Loading Packages

library(tidyverse)
library(lubridate)
library(tidyr)
library(dplyr)
library(ggplot2)
```

```r
# Importing Datasets

library(readr)

daily_activity <- read_csv("/kaggle/input/fitbit/mturkfitbit_export_4.12.16-5.12.16/Fitabase Data 4.12.16-5.12.16/dailyActivity_merged.csv", show_col_types = FALSE)
daily_calories <- read_csv("/kaggle/input/fitbit/mturkfitbit_export_4.12.16-5.12.16/Fitabase Data 4.12.16-5.12.16/dailyCalories_merged.csv", show_col_types = FALSE) 
hourly_calories <- read_csv("/kaggle/input/fitbit/mturkfitbit_export_4.12.16-5.12.16/Fitabase Data 4.12.16-5.12.16/hourlyCalories_merged.csv", show_col_types = FALSE)
hourly_steps <- read_csv("/kaggle/input/fitbit/mturkfitbit_export_4.12.16-5.12.16/Fitabase Data 4.12.16-5.12.16/hourlySteps_merged.csv", show_col_types = FALSE)
sleep_data <- read_csv("/kaggle/input/fitbit/mturkfitbit_export_4.12.16-5.12.16/Fitabase Data 4.12.16-5.12.16/sleepDay_merged.csv", show_col_types = FALSE)
```

## Cleaning

## Cleaning Column Names

```r
#column check for all datasets
colnames(daily_activity)
colnames(sleep_data)
colnames(hourly_calories)
colnames(hourly_steps)
```

```r
names(daily_activity)
names(sleep_data)
names(hourly_calories)
names(hourly_steps)
```

```r
library(dplyr)
library(janitor)

#Clean column names for all datasets
daily_activity <- daily_activity %>% clean_names()
sleep_data <- sleep_data %>% clean_names()
hourly_calories <- hourly_calories %>% clean_names()
hourly_steps <- hourly_steps %>% clean_names()
```

## Convert Date-Time Columns

```r
library(dplyr)
library(lubridate)

# daily_activity: convert activity_date
daily_activity <- daily_activity %>%
  mutate(date = mdy(activity_date))

# sleep_data: convert sleep_day
sleep_data <- sleep_data %>%
  mutate(date = mdy_hms(sleep_day))

# hourly_calories: convert activity_hour
hourly_calories <- hourly_calories %>%
  mutate(activity_hour = mdy_hms(activity_hour),
         date = as_date(activity_hour))

# hourly_steps: convert activity_hour
hourly_steps <- hourly_steps %>%
  mutate(activity_hour = mdy_hms(activity_hour),
         date = as_date(activity_hour))
```

## Check And Remove Duplicates And Null

```r
library(dplyr)

# Daily Activity
daily_activity <- daily_activity %>%
  distinct() %>%          # remove duplicates
  drop_na()               # remove rows with any NA values

# Sleep Data
sleep_data <- sleep_data %>%
  distinct() %>%
  drop_na()

# Hourly Calories
hourly_calories <- hourly_calories %>%
  distinct() %>%
  drop_na()

# Hourly Steps
hourly_steps <- hourly_steps %>%
  distinct() %>%
  drop_na()
```

```r
#Rechecking for null values 
sum(is.na(daily_activity))
sum(is.na(sleep_data))
sum(is.na(hourly_calories))
sum(is.na(hourly_steps))
```

```r
#Rechecking for missing values 
colSums(is.na(daily_activity))
colSums(is.na(sleep_data))
colSums(is.na(hourly_calories))
colSums(is.na(hourly_steps))
```

## Merge Datasets

```r
library(dplyr)
library(lubridate)

daily_sleep <- merge(daily_activity, sleep_data,
                     by = c("id", "date"),
                     all = TRUE)   

hourly_data <- merge(hourly_calories, hourly_steps,
                     by = c("id", "activity_hour", "date"),
                     all = TRUE)
```

# **Step 4 – Analyse**

Average Step Per Day of the Week

```r
library(dplyr)
library(lubridate)

#Add weekday column and calculate average steps
avg_steps_by_day <- daily_activity %>%
  mutate(weekday = wday(date, label = TRUE, abbr = FALSE)) %>%  # Extract weekday name
  group_by(weekday) %>%
  summarise(avg_steps = mean(total_steps, na.rm = TRUE)) %>%
  arrange(match(weekday, c("Sunday", "Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday")))

overall_averages <- daily_activity %>%
  summarise(
    avg_steps = mean(total_steps, na.rm = TRUE),
    avg_calories = mean(calories, na.rm = TRUE)
  )

overall_averages

avg_steps_by_day
```

# **Step 5 – Share**

Data Visualisation and Key Insights

```r
library(ggplot2)

ggplot(daily_activity, aes(x = total_steps, y = calories)) +
  geom_point(alpha = 0.6, colour = "steelblue") +   
  geom_smooth(method = "lm", se = FALSE, colour = "red") + 
  labs(title = "Scatter Plot of Total Steps vs Calories",
       x = "Total Steps",
       y = "Calories Burned") +
  theme_minimal()
```

**Insights**

* Users' average daily steps are below the recommended 10,000 steps. This suggests that most users are not highly active.
* Your scatter plot shows that the number of steps taken and the number of calories burnt are closely linked, but calories burnt can also vary depending on the intensity of activity, not just step count.
* Friday, Saturday, and Sunday tend to show higher activity, while weekdays show lower step counts, likely due to sedentary activities such as sitting at work.
* According to sleep data, many users get less than the recommended 7–8 hours of sleep per night.
* Users are active before and after work because step counts and calorie burn peak in the morning (7–9 am) and evening (5–8 pm).
